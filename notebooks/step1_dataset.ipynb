```python
import pandas as pd
import numpy as np
import os
print("✅ Imports done!")
```

```python
df = pd.read_csv("D:/cyber_threat_detection/data/malicious_phish.csv")
print(f"Columns: {df.columns.tolist()}")
print(f"Shape: {df.shape}")
print(f"\nLabel counts:\n{df['type'].value_counts()}")
df.head(10)
```

```python
df["label"] = df["type"].apply(lambda x: 0 if x == "benign" else 1)
print(f"Total URLs: {len(df)}")
print(f"Safe (0):      {len(df[df['label']==0])}")
print(f"Malicious (1): {len(df[df['label']==1])}")
df = df.drop_duplicates(subset=["url"])
df = df.dropna(subset=["url"])
df = df.reset_index(drop=True)
print(f"\nAfter cleaning: {len(df)}")
os.makedirs("D:/cyber_threat_detection/data", exist_ok=True)
df[["url","label"]].to_csv("D:/cyber_threat_detection/data/raw_urls.csv", index=False)
print("✅ STEP 1 COMPLETE!")
```
