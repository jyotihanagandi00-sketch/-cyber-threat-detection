```python
import pandas as pd
import numpy as np
import shap
import joblib
import matplotlib.pyplot as plt
import pickle
import warnings
warnings.filterwarnings("ignore")
print("✅ Imports done!")
```

```python
rf_model = joblib.load("D:/cyber_threat_detection/models/random_forest.pkl")
df = pd.read_csv("D:/cyber_threat_detection/data/features.csv")
X = df.drop("label", axis=1)
y = df["label"]
X_sample = X.sample(n=500, random_state=42)
print(f"Model loaded ✅ Sample shape: {X_sample.shape}")
```

```python
print("Creating SHAP explainer...")
print("Takes 2-3 minutes...")
explainer   = shap.TreeExplainer(rf_model)
shap_values = explainer.shap_values(X_sample)
print("✅ SHAP values computed!")
```

```python
plt.figure(figsize=(10, 8))
shap.summary_plot(shap_values[:, :, 1], X_sample, plot_type="bar", max_display=15, show=False)
plt.title("Top 15 Features — SHAP Feature Importance", fontsize=14)
plt.tight_layout()
plt.savefig("D:/cyber_threat_detection/data/shap_importance.png", dpi=150, bbox_inches="tight")
plt.show()
print("✅ SHAP plot saved!")
```

```python
with open("D:/cyber_threat_detection/models/shap_explainer.pkl", "wb") as f:
    pickle.dump(explainer, f)
print("✅ SHAP explainer saved!")
print("✅ STEP 6 COMPLETE!")
```
