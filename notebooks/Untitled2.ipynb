```python
import pandas as pd

tranco_df = pd.read_csv(
    "C:/Users/Gazala Anjum B K/Documents/cyber threat detection/cyber-threat-detection/data/top-1m.csv",
    header=None, names=["rank", "domain"]
)
top_10k = tranco_df.head(10000)

modern_urls = pd.DataFrame({
    "url": "https://" + top_10k["domain"],
    "label": 0
})

modern_urls.to_csv(
    "C:/Users/Gazala Anjum B K/Documents/cyber threat detection/cyber-threat-detection/data/modern_urls.csv",
    index=False
)
print(f"Saved {len(modern_urls)} modern benign URLs")
```

```python
import pandas as pd

existing = pd.read_csv(
    "C:/Users/Gazala Anjum B K/Documents/cyber threat detection/cyber-threat-detection/data/raw_urls.csv"
)
modern = pd.read_csv(
    "C:/Users/Gazala Anjum B K/Documents/cyber threat detection/cyber-threat-detection/data/modern_urls.csv"
)

merged = pd.concat([existing, modern], ignore_index=True)
merged = merged.drop_duplicates(subset="url")

merged.to_csv(
    "C:/Users/Gazala Anjum B K/Documents/cyber threat detection/cyber-threat-detection/data/raw_urls_v2.csv",
    index=False
)

print(f"Original: {len(existing)} rows")
print(f"Modern added: {len(modern)} rows")
print(f"Final merged (after dedup): {len(merged)} rows")
print(f"\nLabel distribution:\n{merged['label'].value_counts()}")
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import json

classes = np.array([0, 1])
weights = compute_class_weight(class_weight="balanced", classes=classes, y=merged["label"])
class_weights_v2 = {"0": weights[0], "1": weights[1]}

with open(
    "C:/Users/Gazala Anjum B K/Documents/cyber threat detection/cyber-threat-detection/data/class_weights_v2.json",
    "w"
) as f:
    json.dump(class_weights_v2, f, indent=2)

print(class_weights_v2)
```
